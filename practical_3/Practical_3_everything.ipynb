```python
# Provided inputs
nums = [3, -1, 7, 2, 9, 0, 4]
limit = 4
text = "Room 101: bring 2 apples & 1 banana."

# Global variables
count = 999
summary = "unset"
result = "unset"
```

```python
#First Function
def count_above(seq, lim):
    count = 0
    for valuea in seq:
        if valuea > lim:
            count+=1
    return count

#calling it like this here is just the same as printing atleast as far as I am concerned.
print(count)
print(count_above(nums, limit))
print(count)
#count from the function which we define as 0 doesn't really exist outside of it that is the reason why it remains 999
```

```python
#Second Function

def summarise_text(s):
    summary = {"digits":0, "letters":0, "other":0}
    for character in s:
        if character.isdigit() == True:
            summary["digits"]+= 1
        elif character.isalpha() == True:
            summary["letters"]+= 1
        else: summary["other"]+= 1
    return summary
#It is the same story as above, nothing special to explain
print(summarise_text(text))
print(summary)
```

```python
#Third Function
def aggregate(seq,mode,threshold):
    if mode =="sum":
        result=0
    elif mode =="count":
        result=0
    elif mode == "max":
        result= None
    else:
        print("error not sum, count or max")
    for n in seq:
        if n < 0:
            continue
            #what about a negative threshold? I guess we do not care is the answer
        if n>= threshold:
            if mode =="sum":
                result+= n
            elif mode == "count":
                result+=1
            else:
                if result == None or n > result:
                    result=n
    #struggled a bit with this return and even chopped some parks I liked, do make sure the depth is correct
    return result
print(result)
print(aggregate(nums, "sum", limit))
print(aggregate(nums, "count", limit))
print(aggregate(nums, "max", limit))
print(result)
print(aggregate(nums, "max", 100))
```
